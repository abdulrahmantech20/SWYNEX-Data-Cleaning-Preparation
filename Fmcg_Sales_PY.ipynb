```python
df = df.drop_duplicates()
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'], format='mixed', errors='coerce').bfill().ffill()
```

```python
# Fix: Pass dayfirst=True to specify DD/MM/YYYY format explicitly
df['Order_Date'] = pd.to_datetime(df['Order_Date'], format='mixed', dayfirst=True).dt.strftime('%d/%m/%Y')
```

```python
df['Store_Type'] = df['Store_Type'].str.title().str.strip()
```

```python
print("\n=== CATEGORICAL INCONSISTENCIES (Store_Type) ===")
print(df['Store_Type'].unique())
```

```python
df['Store_Type'] = df['Store_Type'].str.title().str.strip()
```

```python
df['Unit_Price_INR'] = pd.to_numeric(df['Unit_Price_INR'].astype(str).str.replace('₹', '').str.strip(), errors='coerce')
df['Unit_Price_INR'] = df.groupby('Product_Name')['Unit_Price_INR'].transform(lambda x: x.fillna(x.median()))
```

```python
df['Quantity_Sold'] = df['Quantity_Sold'].abs()
df['Payment_Method'] = df['Payment_Method'].fillna('Unknown')
```

```python
df['Delivery_Days'] = df['Delivery_Days'].apply(lambda x: 7 if x > 30 else x)
```

```python
df['Gross_Amount'] = df['Quantity_Sold'] * df['Unit_Price_INR']
df['Discount_Amount'] = df['Gross_Amount'] * (df['Discount_Applied_Pct'] / 100)
df['Net_Sales_INR'] = df['Gross_Amount'] - df['Discount_Amount']
```

```python
print("\n===  ANOMALIES & OUTLIERS ===")
print("Quantity Range (Min to Max):", df['Quantity_Sold'].min(), "to", df['Quantity_Sold'].max())
print("Delivery Days Range (Min to Max):", df['Delivery_Days'].min(), "to", df['Delivery_Days'].max())
```

```python
df.to_csv("cleaned_Fmcg_Sales.csv", index=False)
print("Cleaning complete! File saved successfully.")
```
